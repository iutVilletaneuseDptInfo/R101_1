### Suppression d'un élément à un indice donné

Il est possible de supprimer l'élément à l'indice `i` avec la méthode `pop(i)` qui retourne la valeur supprimée. Par défaut, en l'absence d'argument, `pop()` supprime le dernier élément et retourne la valeur supprimée. 

````python
print("Suppression de l'élément à l'indice 2 :")
valeur = nom.pop(2)
print("Valeur supprimée :", valeur)
print(nom)
````

````python
print("Suppression du dernier élément :")
valeur = nom.pop()
print("Valeur supprimée :", valeur)
print(nom)
````

### Alias

L'instruction suivante ne crée pas une *copie* mais un *alias* de `t`, c'est-à-dire
que `t2` et `t` désignent le même objet jusqu'à ce que l'un d'eux
soit réaffecté.

````python
#Suivre l'exécution avec Python Tutor
t=[0,0,0]
t2 = t     # création d'un alias

t[0] = 1
print(t)
print(t2)

t2[0] = 0
print(t)
print(t2)

t = [2,2,2] # création d'un nouveau tableau à une nouvelle adresse
print(t)
print(t2)
````

### Copie

Pour obtenir un tableau réellement indépendant, et non un simple alias, on utilise la méthode `t.copy()` :

````python
#Suivre l'exécution avec Python Tutor
t = [0, 0, 0]
t2 = t.copy()   # création d'une copie indépendante

t[0] = 1
print(t)        # [1, 0, 0]
print(t2)       # [0, 0, 0]  : t2 n'est pas affecté

t2[1] = 5
print(t)        # [1, 0, 0]  : t n'est pas affecté non plus
print(t2)       # [1, 5, 0]
````

Un *alias* partage le même tableau : le modifier par l'une des deux variables modifie aussi l'autre. Une *copie* (avec `t.copy()`) crée un tableau indépendant : le modifier n'a aucun effet sur l'original.

### Parcourir un tableau

On utilise très souvent des boucles pour *parcourir* un tableau
et agir sur les éléments. <br>
Par exemple, pour afficher uniquement les élément d'indice pair :

````python
i = 0
while i < 7:
    if i%2 == 0 :   # l'indice est pair
        print(vitesse[i])
    i += 1
````

#### Longueur d'un tableau `len()`

L'indice ne doit pas dépasser la taille du tableau sous peine d'être `out of range`
(cf. exemple ci-dessus). <br>
On peut vérifier la taille par la fonction `len()` comme pour les chaînes de caractères

````python
print(len(vitesse))
````

Le programme précédent peut alors être réécrit sans risque d'erreur

````python
i = 0
while i < len(vitesse):
    if i%2 == 0 :
        print(vitesse[i])
    i += 1
````

#### Parcours avec `for ... in ...`

Python propose une syntaxe plus directe pour parcourir un tableau (ou une chaîne de caractères) : la boucle `for ... in ...`, qui donne accès à chaque élément (ou caractère) sans passer par un indice.

````python
# Parcours d'un tableau avec for ... in ...
total = 0
for v in vitesse:
    total += v
print(total)

# Parcours d'une chaîne de caractères
for c in "python":
    print(c)
````

Le mot-clé `in` sert aussi, en dehors d'une boucle, à tester si une valeur *appartient* à un tableau ou à une chaîne : `val in t` renvoie `True` ou `False`.

````python
print(30.2 in vitesse)   # True
print(100 in vitesse)    # False

print("y" in "python")   # True
print("z" in "python")   # False
````

#### Exercice 3 : Occurrences d'une valeur*

Le tableau `notes = [12, 8, 15, 8, 10, 8, 17, 9]` contient des notes d'étudiants. Sans utiliser la méthode `count`, compter et afficher, dans une variable `occurrences`, le nombre de fois où la valeur `8` apparaît dans ce tableau.

````python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
notes = [12, 8, 15, 8, 10, 8, 17, 9]
occurrences = 0

i = 0
while i < len(notes):
    if notes[i] == 8:
        occurrences += 1
    i += 1

print("le nombre d'occurrences est", occurrences)
````

````python
# Test du résultat obtenu
assert occurrences==3
````

#### Exercice 4 : Inversion d'un tableau*

Le tableau `t = [3, 7, 1, 9, 4]` doit être inversé : construire un nouveau tableau `t_inverse` contenant les éléments de `t` dans l'ordre inverse, sans utiliser la méthode `reverse` ni l'écriture `t[::-1]`.

````python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
t = [3, 7, 1, 9, 4]
t_inverse = []

i = len(t) - 1
while i >= 0:
    t_inverse.append(t[i])
    i -= 1

print(t_inverse)
````

````python
# Test du résultat obtenu
assert t_inverse == [4, 9, 1, 7, 3]
````

#### Exercice 8 : Élément majoritaire**

Le tableau `t = [3, 3, 4, 2, 3, 3, 3]` possède un élément majoritaire si une valeur y apparaît strictement plus de la moitié du temps. Sans écrire de fonction, déterminer si `t` possède un élément majoritaire et le placer dans une variable `majoritaire` (`None` s'il n'y en a pas).

````python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
t = [3, 3, 4, 2, 3, 3, 3]

majoritaire = None
i = 0
while i < len(t) and majoritaire is None:
    compteur = 0
    j = 0
    while j < len(t):
        if t[j] == t[i]:
            compteur += 1
        j += 1
    if compteur > len(t) / 2:
        majoritaire = t[i]
    i += 1

if majoritaire is not None:
    print("Élément majoritaire :", majoritaire)
else:
    print("Pas d'élément majoritaire.")
````

````python
# Test du résultat obtenu
assert majoritaire == 3
````
