# <center> Chapitre 12 : Structures imbriquées </center>

Une structure imbriquée est une structure de données (tableau ou dictionnaire) qui contient elle-même des tableaux ou des dictionnaires.
- niveau 2 : un tableau de dictionnaires, un dictionnaire de tableaux, un tableau de tableaux, etc.
- niveau 3 : une structure de niveau 2 elle-même imbriquée dans une autre structure (par exemple, un dictionnaire de tableaux de dictionnaires).

Exemple de niveau 2 (un tableau de dictionnaires) :

```python
nobels_femmes = [
    {"nom": "Maria Ressa", "année": 2021, "domaine": "Paix"},
    {"nom": "Andrea M. Ghez", "année": 2020, "domaine": "Physique"},
]
```

## Accéder à une information

On va du plus général (la structure qui contient) au plus précis, en enchaînant les crochets `[...]` : `variable[...][...]` pour une structure de niveau 2, `variable[...][...][...]` pour une structure de niveau 3.

```python
print(nobels_femmes[0])            # le dictionnaire complet du premier élément
print(nobels_femmes[0]["domaine"]) # une valeur précise de ce dictionnaire
```

## Parcourir une structure imbriquée

On utilise des **boucles imbriquées** : une boucle contenue dans le bloc d'instructions d'une autre boucle. La boucle externe parcourt la structure de plus haut niveau, la boucle interne parcourt chaque sous-structure.

```python
matrice = [[2, 3, 4], [4, 3, 2], [7, 8, 9]]

somme = 0
ligne = 0
while ligne < len(matrice):
    colonne = 0
    while colonne < len(matrice[ligne]):
        somme += matrice[ligne][colonne]
        colonne += 1
    ligne += 1
print(somme)
```

## Boucles imbriquées et fonctions

Encapsuler la boucle interne dans une fonction évite d'écrire des boucles imbriquées explicitement dans l'algorithme principal, et rend le code plus lisible.

```python
def table_mult(i):
    """Affiche la table de multiplication de i.

    :param i: l'entier dont on affiche la table de multiplication.
    :type i: int
    :returns: None
    :rtype: None
    """
    j = 1
    while j <= 10:
        print(i, "*", j, "=", i * j)
        j += 1

i = 1
while i <= 10:
    table_mult(i)
    i += 1
```

## Structures de niveau 3

Une structure de niveau 3 est une structure imbriquée dans une structure elle-même imbriquée. Par exemple, un dictionnaire associant à chaque rayon d'un magasin (la clé) un tableau de produits, chaque produit étant lui-même un dictionnaire :

```python
magasin = {
    "Papeterie": [{"nom": "Cahier", "prix": 2.5, "stock": 30}, {"nom": "Stylo", "prix": 1.2, "stock": 3}],
    "Informatique": [{"nom": "Souris", "prix": 15.0, "stock": 8}],
}

print(magasin["Papeterie"][0]["prix"])   # accès chaîné : dico -> tableau -> dico
```

Pour la parcourir, on imbrique une boucle sur les clés du dictionnaire (ici les rayons) et une boucle sur le tableau associé à chaque clé (ici les produits) :

```python
def valeur_totale(magasin):
    """Calcule la valeur totale du stock d'un magasin, tous rayons confondus.

    :param magasin: dictionnaire associant chaque rayon à son tableau de produits.
    :type magasin: dict
    :returns: la somme de prix * stock pour tous les produits du magasin.
    :rtype: float
    """
    total = 0
    for rayon in magasin:
        for produit in magasin[rayon]:
            total += produit["prix"] * produit["stock"]
    return total

print(valeur_totale(magasin))
```
