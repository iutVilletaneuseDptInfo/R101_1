## Qu'est-ce qu'une fonction ?

Une *fonction* est une **séquence d'instructions** effectuant une tâche précise, un calcul ou un affichage par exemple. <span style="color:red;">Définir</span> une fonction consiste donc à écrire une séquence d'instructions. Une fois qu'une fonction a été définie, on peut <span style="color:red;">appeler la fonction</span>, c'est-à-dire demander l'exécution de la séquence d'instructions associée, aussi souvent qu’on le souhaite et à n’importe quel endroit du programme. 

Les fonctions permettent :
* d'éviter le copier/coller (source d'erreur),
* de structurer le programme et le rendre plus clair,
* de faciliter l'écriture et le test du code.

L'endroit dans le code où la fonction est appelée est dit *programme appelant*. Cela peut être le programme principal ou une autre fonction.
Lors de l'appel d'une fonction : 

* l'exécution du code contenant l'appel de la fonction "se met en pause",
* l'exécution de la fonction commence :
    * **Optionnel :** les paramètres de la fonction, s'il y en a, sont initialisés avec les valeurs données par le programme appelant (on parle de valeurs d'entrée),
    * la séquence d'instructions de la fonction est exécutée,
    * **Optionnel :** la fonction retourne des valeurs (on parle de *valeurs de retour* ou sortie),
    * les *variables locales* de la fonction et les paramètres de la fonction sont détruits,
* l'exécution du code contenant l'appel de la fonction reprend, en utilisant si besoin les valeurs retournées par la fonction.

<img src="img/appel_fonction.png" alt="Appel d'une fonction"  width="700px"/>

**Remarque :** Une fonction peut avoir été définie par un développeur autre que l'auteur du programme appelant. Par exemple, on utilise les fonctions `abs`, `sqrt` et même `print` sans les avoir définies soi-même ! Il est possible de faire cela car la fonction `print` est définie dans langage Python et que les deux autres fonctions sont définies dans le module `math`.

## Fonctions sans paramètre et sans valeur de retour

### Définition

Définir une fonction se fait de la manière suivante :

<img src="img/anatomie_definition_fonction.png" alt="Anatomie de la définition d'une fonction"  width="800px"/>

La première ligne est appelée <span style="color:red;">en-tête</span> ou <span style="color:red;">signature</span> de la fonction et la séquence d'instructions <span style="color:red;">corps</span> de la fonction.

### Appel

L'appel d'une fonction qui ne prend pas de paramètre se fait en indiquant le nom de la fonction suivi de parenthèses ouvrante et fermante.
```python
nom_fonction()
```

### Exemple

Dans une boucle `while` où chaque itération produit de nombreux affichages, il est possible à l'aide d'une fonction de faire apparaître clairement que l'on commence une nouvelle itération. 

````python
def cadre_iteration():
    """Affiche un cadre '===' délimitant le début d'une nouvelle
    itération.

    :returns: None (affiche le cadre).
    :rtype: None
    """
    print("======================")
    print("= Nouvelle itération =")
    print("======================")


i = 0
while i < 5:
    cadre_iteration()
    print("i  =", i)
    print("plein d'affichage...\nBla bla bla\netc")
    i += 1
````

**Remarque :** L'exécution de l'exemple précédent avec Python Tutor montre clairement la pause du code appelant le temps que la fonction s'exécute.

## Fonctions avec paramètres et sans valeur de retour

Une fonction peut avoir besoin d'informations différentes à chaque exécution. Par exemple, une fonction qui affiche un message entouré d'un cadre a besoin de connaître ce message pour fonctionner. On dit qu'elle possède un **paramètre**, et que sa **valeur** est fournie par le code appelant au moment de l'appel.

### Définition

Définir une fonction, c'est notamment préciser son nombre de **paramètres**, c'est-à-dire le nombre de valeurs que l'on doit fournir lors de son appel. Ces paramètres sont indiqués entre parenthèses dans l'en-tête de la fonction.

<img src="img/definition_fonction_parametres.png" alt="Définition d'une fonction avec des paramètres"  width="800px"/>

### Appel

L'appel d'une fonction ayant $k$ paramètres se fait en indiquant le nom de la fonction suivi d'une parenthèse ouvrante, des $k$ valeurs que l'on donne à la fonction et d'une parenthèse fermante.
```python
nom_fonction(val1, val2, ..., valk)
```

Lors de l'appel d'une fonction ayant $k$ paramètres, l'interpréteur commence par créer $k$ variables ayant pour nom les noms des paramètres et pour valeur les valeurs données lors de l'appel. Le premier paramètre est initialisé avec la première valeur, le deuxième avec la deuxième valeur, etc. Il exécute ensuite les instructions du corps de la fonction.

<img src="img/transmission_parametres_fonction.png" alt="Transmission de paramètres à une fonction"  width="800px"/>


### Exemple

La fonction `cadre_iteration` définie précédemment est intéressante, mais on peut la généraliser pour afficher des messages encadrés par n'importe quel symbole. On crée pour cela une fonction `cadre` qui dépend de deux paramètres :

* `message`, le message à afficher encadré ;
* `symbole`, le symbole qui constitue le cadre.

La fonction `cadre` est ainsi beaucoup plus générale que `cadre_iteration`. En contrepartie, il faut préciser à chaque appel le message à afficher et le symbole à utiliser pour le cadre. On décide arbitrairement que la première valeur donnée est le message, et la deuxième le symbole. L'appel `cadre("Nouvelle itération", "=")` doit ainsi produire le même résultat que `cadre_iteration()`.

````python
def cadre(message, symbole):
    """Affiche message encadré par des lignes de caractères symbole.

    :param message: le texte à encadrer.
    :type message: str
    :param symbole: le caractère utilisé pour dessiner le cadre.
    :type symbole: str
    :returns: None (affiche le cadre).
    :rtype: None
    """
    s = ""  # Chaîne vide qui va constituer la ligne de symbole à afficher avant et après le message
    i = 0
    while i < len(message) + 4:  # + 4 car on affiche le symbole, 1 espace, le message, 1 espace, le symbole
        s += symbole
        i += 1

    print(s)
    print(symbole, message, symbole)
    print(s)

cadre("Nouvelle itération", "=")
````

````python
cadre("Oh", "*")
cadre("GEEK", "@")
````

**Attention :** L'ordre dans lequel on donne les valeurs lors de l'appel est important puisque ces valeurs servent à initialiser les paramètres de la fonction. Si l'on change l'ordre, cela n'a plus du tout le même effet : 

````python
# Exemple où l'on se trompe : on donne d'abord le symbole puis 
# le message
cadre("=", "Oups")
````

### Mode de transmission des paramètres

Lorsqu'on appelle une fonction qui attend des paramètres, on lui fournit des **valeurs** d'entrée : des constantes, ou des valeurs stockées dans des variables. Avec les types de valeurs vus jusqu'ici, **appeler une fonction avec une variable ne modifie pas cette variable** : c'est sa valeur qui est transmise à la fonction, pas la variable elle-même.

Voici un exemple illustrant le fait que les variables utilisées pour transmettre des valeurs d'entrée ne sont pas modifiées. L'utilisation de Python Tutor permet de bien comprendre ce qui se passe lors de l'appel.

````python
# Définition de la fonction change()
def change(i):
    """ Modifie localement le paramètre i.

    :param i: valeur passée en paramètre (entier : modification locale).
    :type i: int
    :returns: None (affiche i avant et après modification locale).
    :rtype: None
    """
    print("dans la fonction, i = ", i)
    i = 3
    print("dans la fonction, i = ", i)


# Programme principal
i = 5
print("dans le programme principal, i = ", i)
change(i)
print("dans le programme principal, i = ", i)
````

## Portée des variables

La <span style="color:red;">portée</span> (ou visibilité) d'une variable désigne la partie du code dans laquelle cette variable est accessible. Une variable déclarée dans le corps d'une fonction (ou un paramètre de cette fonction) n'est accessible qu'à l'intérieur de ce corps : on ne peut pas y accéder depuis l'extérieur. On parle alors de <span style="color:red;">portée locale</span> ou de <span style="color:red;">variable locale</span>.

Voici un exemple de code provoquant une erreur, car il tente d'accéder à des variables locales depuis l'extérieur de leur fonction.

````python
# Définition de la fonction affiche_somme
def affiche_somme(x, y):
    """Affiche la somme de x et y.

    :param x: le premier nombre.
    :type x: int
    :param y: le second nombre.
    :type y: int
    :returns: None (affiche la somme de x et y).
    :rtype: None
    """
    z = x + y  # z est une variable locale à la fonction
    print("La somme de", x, "et", y, "vaut", z)


affiche_somme(7, 5)

# Erreur car on accède à une variable locale en dehors du corps 
# de la fonction
print(z)

# Erreur car on accède à un paramètre en dehors du corps de la fonction
print(x)
````

## Docstring

Pour utiliser et appeler  correctement une fonction, il faut savoir ce qu'elle fait, quelles valeurs elle attend en entrée, et dans quel ordre les donner. Pour afficher l'aide d'une fonction déjà définie, il suffit de taper la commande :
```python
?nom_fonction
```
Par exemple, supposons que l'on charge la librairie `turtle`. 

````python
from turtle import *
````

Pour savoir comment s'utilise la fonction `forward`, on tape 

````python
?forward
````

Lorsqu'on définit une fonction, il est important de documenter ce qu'elle fait, afin qu'un·e autre programmeur·se (ou nous-même, plus tard) puisse l'utiliser sans relire son code. On rédige pour cela une chaîne de caractères, délimitée par 3 guillemets (pour pouvoir y insérer des sauts de ligne), que l'on place juste après l'en-tête de la fonction, avant son corps. On appelle cette chaîne une **docstring** (« chaîne de documentation »). C'est elle qui s'affiche lorsqu'on demande l'aide de la fonction avec `?nom_fonction`.

### Le format reST

Pour que la docstring reste claire et exploitable (notamment par des outils qui génèrent automatiquement de la documentation), on suit une convention d'écriture appelée **reST** (reStructuredText). Elle consiste à décrire chaque paramètre et la valeur de retour à l'aide de balises commençant par `:` :

* `:param nom_du_parametre:` — décrit ce que représente le paramètre ;
* `:type nom_du_parametre:` — indique son type (`int`, `str`, `list`, ...) ;
* `:returns:` (ou `:return:`) — décrit la valeur renvoyée par la fonction ;
* `:rtype:` — indique le type de cette valeur de retour.

Une docstring commence en général par une courte phrase résumant ce que fait la fonction, suivie d'une ligne vide puis de ces balises. C'est le format que nous utiliserons pour toutes les fonctions de ce cours.

Voici un exemple de fonction avec docstring :

````python
def cadre(message, symbole):
    """
    Cette fonction affiche un message entouré d'un cadre. 
    Ce cadre est constitué de la répétition d'un caractère.
    Attention, le message doit être sur une seule ligne !

    :param message: message à afficher (chaîne de caractères)
    :type message: str
    :param symbole: symbole utilisé pour faire le cadre (un caractère)
    :type symbole: str
    """
    s = ""  # Chaîne vide qui deviendra la ligne de symbole à afficher 
            # avant et après le message
    i = 0

    # + 4: le symbole, 1 espace au début, 1 espace, le symbole à la fin 
    while i < len(message) + 4:
        s += symbole
        i += 1

    print(s)
    print(symbole, message, symbole)
    print(s)
````

````python
cadre("Oh", "*")
````

**Exercice 6 : Addition unaire***

Un nombre positif ou nul est représenté en base unaire par autant de symboles 0 que la valeur du nombre. Ainsi, la représentation unaire de 5 est : `"00000"`.

1. Définir la fonction `affiche_unaire` prenant en paramètre un nombre et affichant ce nombre en unaire.
2. Définir la fonction `addition_unaire` prenant en paramètre deux nombres et affichant ces deux nombres en base unaire ainsi que leur somme. Par exemple, l'appel de la fonction avec les valeurs 7 et 5 doit donner l'affichage suivant :
```
0000000
+
00000
=
000000000000
```

````python
#------------------#
#   CORRECTION !   #
#------------------#
def affiche_unaire(nb):
    """Affiche l'écriture unaire d'un entier positif donné en base 10.

    :param nb: l'entier positif à convertir et afficher en unaire.
    :type nb: int
    :returns: None (affiche l'écriture unaire, ne la retourne pas).
    :rtype: None
    """
    i = 0
    s = ""
    while i < nb:
        s += "0"
        i += 1
    print(s)
    
def addition_unaire(nb1, nb2):
    """Affiche deux entiers en unaire séparés par un +, 
    puis affiche le résultat de leur addition en unaire.

    :param nb1: le premier entier (en base 10).
    :type nb1: int
    :param nb2: le second entier (en base 10).
    :type nb2: int
    :returns: None (affiche les écritures unaires, ne les retourne pas).
    :rtype: None
    """
    affiche_unaire(nb1)
    print("+")
    affiche_unaire(nb2)
    print("=")
    affiche_unaire(nb1 + nb2)

addition_unaire(7, 5)
````

**Exercice 7 : Addition binaire*****


1. Définir la fonction `affiche_binaire` prenant en paramètre un nombre et affichant ce nombre en binaire.
2. Définir la fonction `addition_binaire` prenant en paramètre deux nombres et affichant ces deux nombres en binaire ainsi que leur somme. Par exemple, l'appel de la fonction avec les valeurs 7 et 5 doit donner l'affichage suivant :
```
111
+
101
=
1100
```

````python
#------------------#
#   CORRECTION !   #
#------------------#

def affiche_binaire(nb):
    """
    fonction qui prend en paramètre un entier positif en base 10
    et affiche son écriture en binaire sous forme d'une chaîne de caractères
    """
    s = ""
    while nb >= 2:
        s = str(nb % 2) + s

        nb //= 2
    s = str(nb) + s
    print(s)


def addition_binaire(nb1, nb2):
    """
    fonction qui prend en paramètres deux entiers en base 10, 
    les affiche en binaire séparés par un + et affiche le résultat de leur addition en binaire
    """
    affiche_binaire(nb1)
    print("+")
    affiche_binaire(nb2)
    print("=")
    affiche_binaire(nb1 + nb2)


addition_binaire(7, 5)
addition_binaire(1, 2)
````
