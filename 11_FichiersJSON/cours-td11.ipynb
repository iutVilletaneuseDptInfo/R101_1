### Le format JSON


Le format **JSON** (JavaScript Object Notation) est aujourd'hui l'un des formats les plus utilisés pour échanger et sauvegarder des données : configuration d'une application, réponse d'un site web, sauvegarde d'une partie de jeu vidéo...

La conversion d'une donnée (comme un dictionnaire Python) en texte JSON s'appelle la **sérialisation** ; l'opération inverse, qui reconstruit la donnée à partir du texte, s'appelle la **désérialisation**. C'est exactement ce que font les fonctions `dump` (sérialiser) et `load` (désérialiser) présentées ci-dessous. 

Bien qu'il soit issu de la notation des objets du langage JavaScript, il est utilisé par la quasi-totalité des langages de programmation, ce qui en fait un format universel pour faire communiquer des programmes entre eux.

Un fichier JSON ne comporte que des associations clés→valeurs (à l'instar des dictionnaires Python). *Une valeur* peut être une autre association clés→valeurs, une liste de valeurs, un entier, un nombre réel, une chaîne de caractères, un booléen ou une valeur nulle. La syntaxe est très proche de celle des dictionnaires Python.

L'exemple suivant est un extrait de fichier de sauvegarde d'un personnage de jeu vidéo. Il contient des **structures imbriquées** : la valeur associée à la clé `"statistiques"` est elle-même un dictionnaire, et celle associée à la clé `"quetes"` est un tableau de dictionnaires. La notion de structures imbriquées est l'objet du prochain chapitre, où elle sera étudiée de manière plus systématique.


```json
{
  "pseudo": "DrakoNeige",
  "niveau": 12,
  "vivant": true,
  "inventaire": ["Épée", "Bouclier", "Potion de soin"],
  "statistiques": {"force": 18, "agilite": 12, "magie": 7},
  "quetes": [
    {"nom": "Forêt Sombre", "terminee": true},
    {"nom": "Tour du Dragon", "terminee": false},
    {"nom": "Grotte Glacée", "terminee": false}
  ]
}
```

**Remarque** : Le format JSON possède ses propres mots-clés pour les booléens et la valeur nulle : `true`, `false` et `null` (en minuscules), différents des mots-clés Python `True`, `False` et `None`. La fonction `load` effectue automatiquement la conversion lors de la lecture : `true` devient `True`, `null` devient `None`, etc.


**Important** : Pour manipuler les fichiers JSON, on utilise le module `json` de Python grâce à l'instruction ```from json import *``` ou ```import json ```.


### Lire un fichier JSON


La fonction `load(f)` retourne le dictionnaire ou le tableau correspondant au fichier JSON représenté par la variable de flux `f`.

>|  |  |
>| :-- | :-- |
>| `load(f)` | Retourne  le **dictionnaire** correspondant au fichier JSON représenté par la variable de flux `f`.|

**Remarque** : Les clefs et les valeurs du dictionnaire créé par la fonction `load` ne sont pas nécessairement des chaînes de caractères.

Voici un exemple d'utilisation de cette fonction, à partir du fichier `files/personnage.json` reproduit ci-dessus.


````python
#1ère possibilité
from json import *
fichier = open ("./files/personnage.json","r",encoding='utf-8')
personnage = load(fichier)
fichier.close()
print (personnage)

print(type(personnage))
````

````python
#2ème possibilité
import  json
fichier = open ("./files/personnage.json","r",encoding='utf-8')
personnage = json.load(fichier)
fichier.close()
print (personnage)
````

Ainsi il est possible d'accéder aux éléments du dictionnaire comme suit : 


````python
print(personnage['pseudo'])
````

````python
print(personnage['niveau'])
````

````python
quetes = personnage['quetes']
print(quetes)
````

Comme le résultat de la variable `quetes` est un tableau, il est possible d'accéder aux éléments en utilisant l'accès direct ou en utilisant les boucles. 


````python
premiere_quete = quetes[0]
print(premiere_quete)
print(len(premiere_quete))
````

````python
i=0
while (i<len(quetes)):
    print(quetes[i])
    i=i+1
````

Cette écriture est équivalente à une écriture un peu plus complexe : 


````python
print (personnage['quetes'][0])
````

Puisque le résultat `premiere_quete` est également un dictionnaire, alors il est possible d'accéder aux éléments : 


````python
print (premiere_quete['nom'])
print (premiere_quete['terminee'])
````

### Écrire dans un fichier JSON


La fonction `dump(dictionnaire, fichier, indent=n, ensure_ascii=False)` permet d'enregistrer en texte au format JSON un dictionnaire ou un tableau directement dans le fichier `fichier` précédemment ouvert, avec une éventuelle indentation. Les paramètres autres que `dictionnaire` et `fichier`sont facultatifs.

L'option `ensure_ascii` vaut par défaut `True`, les caractères non-ASCII sont alors mal interprétés. Si `ensure_ascii=False`, les caractères non-ASCII sont restitués tels qu'ils sont.


````python
from json import *
personnage2 = {'pseudo': 'Lyra',
 'niveau': 7,
 'vivant': True,
 'inventaire': ['Arc', 'Carquois', 'Herbes medicinales'],
 'statistiques': {'force': 9, 'agilite': 19, 'magie': 14},
 'quetes': [{'nom': 'Marais Trouble', 'terminee': True}]}
````

````python
fichier2 = open ("./files/personnage2.json","w", encoding='utf-8')

dump(personnage2, fichier2, indent=4, ensure_ascii=False)

fichier2.close()
````

### Exemple : sauvegarder et recharger des données

Un usage très courant du format JSON consiste à **sauvegarder** une donnée dans un fichier pour pouvoir la **recharger** plus tard, par exemple d'une exécution du programme à l'autre. 

On illustre ce principe avec un petit carnet de contacts, associant des noms à des numéros de téléphone.


````python
from json import dump, load

def sauvegarder_carnet(carnet, chemin):
    """Enregistre un carnet de contacts dans un fichier JSON.

    :param carnet: le dictionnaire nom -> numéro de téléphone à enregistrer.
    :type carnet: dict
    :param chemin: le chemin du fichier JSON de destination.
    :type chemin: str
    :returns: None
    :rtype: None
    """
    f = open(chemin, 'w', encoding='utf-8')
    dump(carnet, f, indent=2, ensure_ascii=False)
    f.close()
    
carnet = {"Alice": "0601020304", "Bob": "0605060708"}
sauvegarder_carnet(carnet, "files/carnet.json")
````

````python
def charger_carnet(chemin):
    """Recharge un carnet de contacts depuis un fichier JSON.

    :param chemin: le chemin du fichier JSON à charger.
    :type chemin: str
    :returns: le dictionnaire nom -> numéro de téléphone chargé.
    :rtype: dict
    """
    f = open(chemin, 'r', encoding='utf-8')
    carnet = load(f)
    f.close()
    return carnet

print(charger_carnet("files/carnet.json"))
````

````python
def test_carnet():
    """Vérifie qu'un carnet sauvegardé puis rechargé est identique à l'original.

    :returns: None (échoue par assertion en cas d'incohérence).
    :rtype: None
    """
    carnet_original = {"Alice": "0601020304", "Bob": "0605060708"}
    sauvegarder_carnet(carnet_original, "files/carnet.json")
    recharge = charger_carnet("files/carnet.json")
    assert recharge == carnet_original
    print("Test de sauvegarder_carnet et charger_carnet : ok")

test_carnet()
````

### Exemple : calculer une statistique à partir d'un fichier JSON

Un fichier JSON peut aussi être utilisé comme simple source de données, à partir de laquelle on calcule une statistique. Le fichier `files/prix_livres.json` contient un dictionnaire associant à chaque titre de livre son prix.


````python
from json import load
from math import isclose

def prix_moyen(chemin):
    """Calcule le prix moyen des livres à partir d'un fichier JSON.

    :param chemin: le chemin du fichier JSON associant chaque titre à son prix.
    :type chemin: str
    :returns: le prix moyen des livres.
    :rtype: float
    """
    f = open(chemin, 'r', encoding='utf-8')
    livres = load(f)
    f.close()
    total = 0
    for titre in livres:
        total += livres[titre]
    return total / len(livres)

print(prix_moyen("files/prix_livres.json"))
````

````python
def test_prix_moyen():
    """Teste unitairement prix_moyen sur le fichier fourni.

    :returns: None (échoue par assertion si le résultat est incorrect).
    :rtype: None
    """
    moyenne = prix_moyen("files/prix_livres.json")
    assert isclose(moyenne, 8.89)
    print("Test de la fonction prix_moyen : ok")

test_prix_moyen()
````
