## Tableau en paramètre d'une fonction

Une fonction peut prendre un tableau en paramètre. Cela permet
d'appliquer un même traitement à plusieurs tableaux.

**Exemple** :

````python
def affiche_premier1(tab):
    """Affiche le premier élément du tableau tab passé en paramètre.

    :param tab: le tableau dont on veut afficher le premier élément.
    :type tab: list
    :returns: None (affiche le premier élément du tableau).
    :rtype: NoneType

    :Exemples:
    >>> affiche_premier1([1, 2, 3])
    Premier élément : 1
    >>> affiche_premier1([ "Bonjour", "Lundi"])
    Premier élément : Bonjour
    """
    print("Premier élément : ",tab[0])
    
t = ["oui", "non", "peut-être"]
t2 = [34, 22, 6, 70]

affiche_premier1(t)
affiche_premier1(t2)
````

**Remarque :** Passer un tableau vide à cette fonction provoque une
erreur `out of range` car on lit un élément inexistant.

````python
t = []
affiche_premier1(t)
````

Pour éviter ces problèmes d'exécution, il faut tester si le tableau est vide,  et ne pas chercher le premier élément dans ce cas. Le programme peut également afficher un message d'erreur pour avertir l'utilisateur.

````python
def affiche_premier2(tab):
    """Affiche le premier élément du tableau tab, ou un message si tab est vide.

    :param tab: le tableau dont on veut afficher le premier élément.
    :type tab: list
    :returns: None (affiche le premier élément du tableau, ou un message d'erreur s'il est vide).
    :rtype: NoneType

    :Exemples:
    >>> affiche_premier2([ 1, 2, 3])
    Premier élément : 1
    >>> affiche_premier2([])
    Désolé, le tableau est vide !
    """
    if len(tab) == 0:              # teste si le tableau est vide
        print("Désolé, le tableau est vide !")
    else:
        print("Premier élément : ", tab[0])
        

affiche_premier2(t)
````

## Tableau en valeur de retour d'une fonction

Une fonction peut  renvoyer un tableau.

**Exemple** :

````python
from random import *

def tab_alea():
    """Construit un tableau contenant 2 valeurs entières aléatoires.

    La première est comprise entre 1 et 20, la seconde entre 1 et 100.

    :returns: le tableau contenant les 2 entiers aléatoires.
    :rtype: list

    :Exemple:
    >>> tab_alea()
    [3, 64]
    """
    t = []
    t.append(randint(1,20))
    t.append(randint(1,100))
    return t

t0=tab_alea()
print(t0)
````

````python
def test_tab_alea():
    """Teste unitairement la fonction tab_alea (bornes des deux valeurs tirées).

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    t = tab_alea()
    print(t)
    assert len(t) == 2, "le tableau doit contenir 2 valeurs !"
    assert t[0] >= 1 and t[0] <= 20, "1ere valeur entre 1 et 20 ; "
    assert t[1] >= 1 and t[1] <= 100, "2eme valeur entre 1 et 100 !"
    print("Test de la fonction tab_alea : ok")
    
test_tab_alea()
````

### Exemple d'affichage

Afficher un seul élément précédé d'un tiret sur chaque ligne : la fonction ne renvoie rien.

````python
def affiche_tirets(t):
    """Affiche les éléments d'un tableau, chacun sur une ligne, précédé d'un tiret.

    :param t: le tableau contenant les éléments à afficher.
    :type t: list
    :returns: None (affiche les éléments du tableau).
    :rtype: NoneType

    :Exemple:
    >>> affiche_tirets([1,2,"toto"])
    - 1
    - 2
    - toto
    """
    i=0
    while i < len(t):
        print (" - ", t[i])
        i+=1
        
t = ["oui", "non", "peut-être"]
affiche_tirets(t)
````

### Exemple de test

Tester si tous les éléments sont nuls : la fonction renvoie un booléen.

````python
def tous_nuls(t):
    """Teste si tous les éléments d'un tableau sont égaux à l'entier 0.

    :param t: le tableau des valeurs à tester.
    :type t: list
    :returns: True si tous les éléments sont nuls, False sinon.
    :rtype: bool

    :Exemple:
    >>> tous_nuls([0,0,0])
    True
    >>> tous_nuls([0,1,0])
    False
    """
    i=0
    while i < len(t) and t[i] == 0:
        i+=1
    return i==len(t)

def test_tous_nuls():
    """Teste unitairement la fonction tous_nuls sur plusieurs cas.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    assert tous_nuls([0,0,0])
    assert not tous_nuls([0,0,1])
    assert tous_nuls([])
    print("Test de la fonction tous_nuls : ok")

test_tous_nuls()
````

### Exemple de somme ou dénombrement

Calculer la somme des éléments : la fonction renvoie un nombre.

````python
def somme_tab(t):
    """Retourne la somme des éléments d'un tableau contenant des valeurs numériques.

    :param t: le tableau dont on veut sommer les éléments.
    :type t: list
    :returns: la somme des éléments du tableau.
    :rtype: int ou float

    :Exemple:
    >>> somme_tab([1,2,3])
    6
    >>> somme_tab([])
    0
    """
    somme = 0
    i=0
    while i < len(t):
        somme += t[i]
        i+=1
    return somme

def test_somme_tab():
    """Teste unitairement la fonction somme_tab sur plusieurs cas.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    assert somme_tab([]) == 0
    assert somme_tab([1]) == 1
    assert somme_tab([-1,1]) == 0
    assert somme_tab([1,2,3]) == 6
    print("Test de la fonction somme_tab : ok")

test_somme_tab()
````

### Exemples de modification du paramètre

Les fonctions modifiant des tableaux ne renvoient souvent rien : elles modifient directement le tableau passé en paramètre. En fait, c'est l'adresse du tableau qui est transmise à la fonction, ce qui permet de modifier ce qui est stocké en mémoire.

**$1^{er}$ exemple : Ajouter une valeur `val` à tous les éléments du tableau**

````python
def ajoute(t,val):
    """Ajoute une valeur à tous les éléments d'un tableau, sur place.

    :param t: le tableau auquel on souhaite ajouter la valeur.
    :type t: list
    :param val: la valeur à ajouter à tous les éléments du tableau.
    :type val: int ou float
    :returns: None (t est modifié sur place).
    :rtype: NoneType

    :Exemples:
    >>> ajoute([1, 2,10],3)
    >>> ajoute([],3)
    >>> ajoute([1],-1)
    """
    i=0
    while i < len(t):
        t[i] += val
        i+=1

def test_ajoute():
    """Teste unitairement la fonction ajoute sur plusieurs cas.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    t = [1, 2,10]
    ajoute(t,3)
    assert t == [4, 5, 13]
    
    t = []
    ajoute(t,3)
    assert t == []
    
    t = [1]
    ajoute(t,-1)
    assert t == [0]
    print("Test de la fonction ajoute : ok")
        
test_ajoute()
````

**$2^{ème}$ exemple : Insérer un élément en avant-dernière position**

````python
def insere_avant_dernier(t, elt):
    """Insère un élément à l'avant-dernière position d'un tableau, tous deux passés en paramètre.

    :param t: le tableau auquel on souhaite ajouter un élément en avant-dernière position.
    :type t: list
    :param elt: l'élément à ajouter.
    :type elt: n'importe quel type
    :returns: None (t est modifié sur place).
    :rtype: NoneType

    :Exemple:
    >>> insere_avant_dernier([1,3], 2)
    """
    n = len(t)
    tmp = t[n-1]
    t[n-1] = elt
    t.append(tmp)
    

def test_insere_avant_dernier():
    """Teste unitairement la fonction insere_avant_dernier sur plusieurs cas.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    t = [1,3]
    insere_avant_dernier(t, 2)
    assert t == [1, 2, 3]
    
    t = ["a", 3]
    insere_avant_dernier(t, ":")
    assert t == ["a", ":", 3]
    print("Test de la fonction insere_avant_dernier : ok")
    
test_insere_avant_dernier()
````

**$3^{ème}$ exemple: Remettre à zéro tout un tableau**

**Remarque** : pour modifier un tableau dans une fonction, il faut
affecter les éléments concernés un à un et non faire une affectation
globale.

````python
def mise_a_zero(t):
    """Remplace, dans le tableau passé en paramètre, tous les éléments par la valeur 0.

    :param t: le tableau dont on veut mettre le contenu à zéro, sur place.
    :type t: list
    :returns: None (t est modifié sur place).
    :rtype: NoneType

    :Exemple:
    >>> mise_a_zero([1, "toto", 3])
    """
    i=0
    while i < len(t):
        t[i] = 0
        i+=1

        
def test_mise_a_zero():
    """Teste unitairement la fonction mise_a_zero sur plusieurs cas.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    t = [1, "toto"]
    mise_a_zero(t)
    assert t == [0, 0]
    
    t = []
    mise_a_zero(t)
    assert t == []
    
    print("Test de la fonction mise_a_zero : ok")

test_mise_a_zero()
````

L'affectation globale crée une variable locale qui est détruite à la sortie
de la fonction ; le tableau passé en paramètre n'est donc pas modifié.

````python
def affectation(t):
    """Illustre la réaffectation locale d'un paramètre : sans effet sur le tableau de l'appelant.

    :param t: le tableau reçu en paramètre (réaffecté localement à un nouvel objet).
    :type t: list
    :returns: None (affiche la copie locale après réaffectation).
    :rtype: NoneType

    :Exemple:
    >>> affectation([1,2])
    local :  [0, 0, 0]
    """
    t = [0,0,0]
    print("local : ",t)
    
t = [1,1,1]
print("avant l'appel",t)
affectation(t)
print("après l'appel : ",t)
````

### Exemple de création d'un tableau

La fonction renvoie le tableau créé. Si nécessaire, on précise la taille voulue en paramètre. La taille peut aussi être saisie dans la fonction dans ce cas aucun paramètre n'est passé à la fonction.

````python
def multiples_de_10(n):
    """Retourne un tableau contenant les n premiers multiples de 10.

    :param n: le nombre de multiples de 10 à produire (la longueur du tableau).
    :type n: int
    :returns: le tableau contenant les n premiers multiples de 10.
    :rtype: list

    :Exemples:
    >>> multiples_de_10(0)
    []
    >>> multiples_de_10(1)
    [0]
    >>> multiples_de_10(3)
    [0, 10, 20]
    """
    t=[]
    i=0
    while i<n:
        t.append(10*i)
        i+=1
    return t

def test_multiples_de_10():
    """Teste unitairement la fonction multiples_de_10 sur plusieurs cas.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    assert multiples_de_10(0) == []
    assert multiples_de_10(1) == [0]
    assert multiples_de_10(3) == [0, 10, 20]
    print("Test de la fonction multiples_de_10 : ok")

test_multiples_de_10()
````
