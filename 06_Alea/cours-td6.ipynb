Comment un ordinateur, qui est une machine déterministe, peut-il tirer des nombres « au hasard » ? Un algorithme appliqué aux mêmes paramètres donne toujours le même résultat. Pourtant, certaines opérations sont suffisamment imprévisibles pour donner des résultats qui semblent aléatoires. Les nombres obtenus sont dits *pseudo-aléatoires*.

Plus précisément, on utilise une suite qui contient des centaines de milliers de nombres, chacun étant calculé en fonction du précédent, ce qui laisse penser, en les regardant, qu'ils sont aléatoires. Pour cela, ils satisfont des propriétés statistiques de répartition (si les 3/4 des nombres étaient pairs, par exemple, on n'aurait pas une suite pseudo-aléatoire). Par abus de langage, on dira que ces nombres sont aléatoires.

*Remarque :* le principe des suites pseudo-aléatoires est commun à tous les langages de programmation, qui n'utilisent toutefois pas forcément la même suite.

### Génération aléatoire d'entiers

La fonction `randint(a, b)` retourne un entier aléatoire compris entre `a` et `b` inclus.

Par exemple, pour afficher 3 entiers entre 0 et 100 (inclus), on peut écrire

```python
print("1er nombre entre 0 et 100 (inclus) : ", randint(0,100))
print("2e nombre entre 0 et 100 (inclus) : ", randint(0,100))
print("3e nombre entre 0 et 100 (inclus) : ", randint(0,100))
```

et pour afficher un nombre aléatoire entre -20 et 20 (inclus) :

```python
print("nombre entre -20 et 20 : ", randint(-20,20))
```

### Génération aléatoire de flottants

La fonction `random()` permet d'engendrer un nombre flottant aléatoire de l'intervalle $[0, 1[$, *selon une loi uniforme* : chaque sous-intervalle de même longueur a la même chance d'être atteint (par exemple, il y a autant de chances d'obtenir une valeur dans $[0.1, 0.2]$ que dans $[0.7, 0.8]$).

Pour tirer un nombre de l'intervalle $[a, b[$, il suffit de multiplier le résultat de `random()` par la taille de l'intervalle, soit $b - a$, puis de décaler l'intervalle en fonction de la borne inférieure, c'est-à-dire d'ajouter $a$.

Ainsi, le code suivant affiche 3 flottants aléatoires : le premier compris entre 0 et 1, le second entre 0 et 10, et le troisième entre 5 et 20.

```python
print("Nombre flottant entre 0 et 1 : ", random())
print("Nombre flottant entre 0 et 10 : ", 10*random())
print("Nombre flottant entre 5 et 20 : ", 15*random() + 5)
```

On peut aussi définir une fonction qui fait cela pour un intervalle $[a, b[$ quelconque :

```python
def rand_entre(a, b):
    """Retourne un nombre réel aléatoire compris entre a et b.

    :param a: la borne inférieure.
    :type a: int ou float
    :param b: la borne supérieure.
    :type b: int ou float
    :returns: un nombre réel aléatoire dans l'intervalle [a, b[.
    :rtype: float
    """
    return random()*(b-a) + a

print(rand_entre(15,20))
```

```python
def test_rand_entre():
    """Teste que le résultat reste toujours dans les bornes attendues."""
    for _ in range(50):
        assert 15 <= rand_entre(15, 20) <= 20
    print("Test de la fonction rand_entre : ok")

test_rand_entre()
```

Cette fonction est en fait prédéfinie en Python sous le nom d'`uniform(a, b)`, mais cela fait plutôt figure d'exception.

### Différentes générations de nombres aléatoires

Pour s'assurer de ne pas toujours commencer au même endroit de la suite pseudo-aléatoire, et donc de ne pas tirer les mêmes valeurs à chaque exécution, on utilise un entier appelé *graine*. La graine sert à déterminer le nombre de départ dans la suite utilisée par la fonction `random()`. Il faut la modifier à chaque exécution du programme ; pour cette raison, elle est souvent engendrée à partir de l'état du système, par exemple en lisant la valeur de l'horloge. En Python, la graine est engendrée automatiquement lors de l'exécution d'un programme qui utilise des fonctions de la bibliothèque `random`.
