Comment un ordinateur, qui est une machine déterministe, peut-il tirer des nombres « au hasard » ? Un algorithme appliqué aux mêmes paramètres donne toujours le même résultat. Pourtant, certaines opérations sont suffisamment imprévisibles pour donner des résultats qui semblent aléatoires. Les nombres obtenus sont dits *pseudo-aléatoires*.

Plus précisément, on utilise une suite qui contient des centaines de milliers de nombres, chacun étant calculé en fonction du précédent, ce qui laisse penser, en les regardant, qu'ils sont aléatoires. Pour cela, ils satisfont des propriétés statistiques de répartition (si les 3/4 des nombres étaient pairs, par exemple, on n'aurait pas une suite pseudo-aléatoire). Par abus de langage, on dira que ces nombres sont aléatoires.

*Remarque :* le principe des suites pseudo-aléatoires est commun à tous les langages de programmation, qui n'utilisent toutefois pas forcément la même suite.

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
def plusieurs_des(n):
    """Simule le lancer de n dés à 6 faces et renvoie la somme des résultats.

    :param n: le nombre de dés lancés.
    :type n: int
    :returns: la somme des n résultats, chacun compris entre 1 et 6.
    :rtype: int
    """
    somme = 0
    i = 0
    while i < n:
        somme += lancer()
        i += 1
    return somme

print("Lancer de deux dés :", plusieurs_des(2))
print("Lancer de trois dés :", plusieurs_des(3))
```

### Exercice 2 : Soldes aléatoires**

Pour les soldes, un magasin propose une formule originale. Lors du paiement en caisse, une réduction de 30, 50 ou 70 % est appliquée de façon aléatoire.

#### Question 1 : Calcul du taux aléatoire

Écrire une fonction `taux()` retournant 30, 50 ou 70, la valeur étant tirée au hasard.

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
def taux():
    """Tire au hasard un taux de réduction parmi 30, 50 et 70 (en %).

    :returns: 30, 50 ou 70, chacun avec la même probabilité.
    :rtype: int
    """
    de = randint(1, 3)  # on tire un entier entre 1 et 3
    if de == 1:
        reduction = 30
    elif de == 2:
        reduction = 50
    else:
        reduction = 70
    return reduction

print(taux())
```

Une variante plus concise, mais moins générale, de la même fonction :

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
def taux2():
    """Variante concise de taux(), équivalente mais moins générale.

    :returns: 30, 50 ou 70, chacun avec la même probabilité.
    :rtype: int
    """
    return 10 + randint(1, 3) * 20

print(taux2())
```

```python
def test_taux():
    """Teste que le taux renvoyé est toujours l'une des trois valeurs attendues."""
    for i in range(50):
        assert taux() in (30, 50, 70)
    print("Test de la fonction taux : ok")

test_taux()
```

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
def prix_reduit(prix, reduction):
    """Calcule le prix à payer après application d'une réduction.

    :param prix: le prix initial.
    :type prix: float
    :param reduction: le taux de réduction, en pourcentage.
    :type reduction: int ou float
    :returns: le prix après réduction.
    :rtype: float
    """
    return prix * (100 - reduction) / 100

print("Prix ? ")
prix = float(input())
reduction = taux()
print("Réduction", reduction, "%. Prix à payer", prix_reduit(prix, reduction), "euros")
```

### Exercice 3 : Jeu du fakir**

Dans le jeu du fakir, l'utilisateur doit trouver un nombre mystère compris entre 1 et 100 en au plus 6 essais (7 essais sont toujours suffisants). À chaque essai, le programme indique si le nombre mystère est plus petit ou plus grand que la valeur saisie, ou que l'utilisateur a trouvé. Écrire le programme correspondant.

**Indication :** on commencera par écrire une fonction `indice(essai, mystere)` qui compare une proposition au nombre mystère, ce qui permet de la tester indépendamment du reste du jeu.

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#

def indice(essai, mystere):
    """Compare une proposition au nombre mystère.

    :param essai: le nombre proposé par le joueur.
    :type essai: int
    :param mystere: le nombre mystère à deviner.
    :type mystere: int
    :returns: -1 si essai est trop petit, 1 si essai est trop grand, 
               0 si trouvé.
    :rtype: int
    """
    if essai < mystere:
        return -1
    elif essai > mystere:
        return 1
    else:
        return 0
```

```python
def test_indice():
    """Teste la fonction indice sur des cas simples."""
    assert indice(5, 10) == -1
    assert indice(15, 10) == 1
    assert indice(10, 10) == 0
    print("Test de la fonction indice : ok")

test_indice()
```

Le jeu proprement dit s'appuie sur `indice` pour piloter la boucle des essais :

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
nb_myst = randint(1, 100)
essai = -1
n_essai = 0
while essai != nb_myst and n_essai < 6:
    n_essai += 1
    print("Tentez votre chance, trouvez le nombre magique entre 1 et 100")
    essai = int(input())
    resultat = indice(essai, nb_myst)
    if resultat == -1:
        print("Le nombre magique est plus grand")
    elif resultat == 1:
        print("Le nombre magique est plus petit")

if essai == nb_myst:
    print("Vous avez trouvé le nombre magique en", n_essai,
          "coups, vous avez un don médiumnique.")
else:
    print("Perdu, la bonne réponse était", nb_myst)
```

**Autre version**, qui remplace le test sur `essai` par un booléen `pasTrouve` :

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
nb_myst = randint(1, 100)
pasTrouve = True
n_essai = 0

while n_essai < 6 and pasTrouve:
    n_essai += 1
    print("Tentez votre chance, trouvez le nombre magique entre 1 et 100")
    essai = int(input())
    resultat = indice(essai, nb_myst)
    if resultat == -1:
        print("Le nombre magique est plus grand")
    elif resultat == 1:
        print("Le nombre magique est plus petit")
    else:
        print("c'est gagné !")
        pasTrouve = False
```

### Génération aléatoire de flottants

La fonction `random()` permet d'engendrer un nombre flottant aléatoire de l'intervalle $[0, 1[$, *selon une loi uniforme* : chaque sous-intervalle de même longueur a la même chance d'être atteint (par exemple, il y a autant de chances d'obtenir une valeur dans $[0.1, 0.2]$ que dans $[0.7, 0.8]$).

Pour tirer un nombre de l'intervalle $[a, b[$, il suffit de multiplier le résultat de `random()` par la taille de l'intervalle, soit $b - a$, puis de décaler l'intervalle en fonction de la borne inférieure, c'est-à-dire d'ajouter $a$.

Ainsi, le code suivant affiche 3 flottants aléatoires : le premier compris entre 0 et 1, le second entre 0 et 10, et le troisième entre 5 et 20.

```python
print("Nombre flottant entre 0 et 1 : ", random())
print("Nombre flottant entre 0 et 10 : ", 10*random())
print("Nombre flottant entre 5 et 20 : ", 15*random() + 5)
```

On peut aussi définir une fonction qui fait cela pour un intervalle $[a, b[$ quelconque :

```python
def rand_entre(a, b):
    """Retourne un nombre réel aléatoire compris entre a et b.

    :param a: la borne inférieure.
    :type a: int ou float
    :param b: la borne supérieure.
    :type b: int ou float
    :returns: un nombre réel aléatoire dans l'intervalle [a, b[.
    :rtype: float
    """
    return random()*(b-a) + a

print(rand_entre(15,20))
```

```python
def test_rand_entre():
    """Teste que le résultat reste toujours dans les bornes attendues."""
    for i in range(50):
        assert 15 <= rand_entre(15, 20) <= 20
    print("Test de la fonction rand_entre : ok")

test_rand_entre()
```

Cette fonction est en fait prédéfinie en Python sous le nom d'`uniform(a, b)`, mais cela fait plutôt figure d'exception.

### Différentes générations de nombres aléatoires

Pour s'assurer de ne pas toujours commencer au même endroit de la suite pseudo-aléatoire, et donc de ne pas tirer les mêmes valeurs à chaque exécution, on utilise un entier appelé *graine*. La graine sert à déterminer le nombre de départ dans la suite utilisée par la fonction `random()`. Il faut la modifier à chaque exécution du programme ; pour cette raison, elle est souvent engendrée à partir de l'état du système, par exemple en lisant la valeur de l'horloge. En Python, la graine est engendrée automatiquement lors de l'exécution d'un programme qui utilise des fonctions de la bibliothèque `random`.

#### Question 1 : Tirage d'un rectangle aléatoire

Écrire une fonction `rectangle_aleatoire()`, sans paramètre, qui tire au hasard les dimensions d'un rectangle : la largeur uniformément entre 3 et 4, et la longueur uniformément entre 7 et 9. La fonction renvoie le couple `(largeur, longueur)`.

Quelles sont l'aire minimale et l'aire maximale possibles pour un tel rectangle ?

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
def rectangle_aleatoire():
    """Tire au hasard les dimensions d'un rectangle.

    La largeur est tirée uniformément dans [3, 4[ et la longueur
    uniformément dans [7, 9[.

    :returns: le couple (largeur, longueur).
    :rtype: tuple
    """
    largeur = random() + 3
    longueur = 2 * random() + 7
    return largeur, longueur
```

```python
def test_rectangle_aleatoire():
    """Teste que les dimensions tirées restent dans les bornes attendues."""
    for i in range(50):
        l, L = rectangle_aleatoire()
        assert 3 <= l <= 4
        assert 7 <= L <= 9
    print("Test de la fonction rectangle_aleatoire : ok")

test_rectangle_aleatoire()
```

### Exercice 5 : Fléchettes**

On envoie des fléchettes sur une cible de rayon 18. Les coordonnées $x$ et $y$ des impacts sont tirées au hasard entre -20 et 20, l'origine étant le centre de la cible. On note $d = \sqrt{x^2 + y^2}$ la distance au centre et on marque :

- 1 point si $d$ est compris entre 12 et 18,
- 2 points si $d$ est compris entre 6 et 12,
- 5 points si $d$ est compris entre 2 et 6,
- 10 points si $d$ est inférieur à 2 (« dans le mille »),
- 0 point si on rate la cible.

#### Question 1 : Score d'un impact

Écrire une fonction `score_flechette(x, y)` qui prend en paramètre les coordonnées `x` et `y` d'un impact, et renvoie le score obtenu (0, 1, 2, 5 ou 10 points) selon la distance au centre.

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
from math import sqrt


def score_flechette(x, y):
    """Calcule le score obtenu pour un impact de coordonnées (x, y).

    :param x: l'abscisse de l'impact.
    :type x: float
    :param y: l'ordonnée de l'impact.
    :type y: float
    :returns: le score (0, 1, 2, 5 ou 10 points) selon la distance au centre.
    :rtype: int
    """
    d = sqrt(x * x + y * y)
    if d < 2:
        return 10
    elif d < 6:
        return 5
    elif d < 12:
        return 2
    elif d < 18:
        return 1
    else:
        return 0
```

```python
def test_score_flechette():
    """Teste le score pour des distances caractéristiques au centre."""
    assert score_flechette(0, 0) == 10
    assert score_flechette(4, 0) == 5
    assert score_flechette(0, 10) == 2
    assert score_flechette(15, 0) == 1
    assert score_flechette(19, 0) == 0
    print("Test de la fonction score_flechette : ok")

test_score_flechette()
```

#### Question 2 : Simulation du jeu

En utilisant `score_flechette`, écrire un programme qui simule des lancers de fléchettes à coordonnées aléatoires, affiche le résultat de chaque lancer ainsi que le score cumulé, et demande à chaque tour si l'utilisateur veut rejouer.

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
jeu = True
score = 0
while jeu:
    ch = "Vous lancez...    "
    x = 40 * random() - 20
    y = 40 * random() - 20
    pts = score_flechette(x, y)
    score += pts
    if pts == 10:
        print(ch + "   Dans le mille !!   10 pts")
    elif pts == 5:
        print(ch + "-->  5 pts    bravo!")
    elif pts == 2:
        print(ch + "-->  2 pts    c'est pas mal")
    elif pts == 1:
        print(ch + "-->  1 pt     peut mieux faire")
    else:
        print(ch + "   Raté !!   Vous avez besoin de lunettes")
    print("  *** Score : ", score)
    choix = input("Rejouer ? (0/n) ")
    if choix == "n" or choix == "N":
        jeu = False
```

### Exercice 6 : Marche aléatoire*

On simule une marche aléatoire sur une ligne graduée : à chaque étape, une pièce est tirée au hasard, 
- si c'est pile, on avance de 1, 
- si c'est face, on recule de 1. 

1. Écrire une fonction `marche(n)` qui simule `n` étapes à partir de la position 0 et renvoie la position finale. 

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
def marche(n):
    """Simule une marche aléatoire à une dimension sur n étapes, 
       en partant de la position 0.

    :param n: le nombre d'étapes de la marche.
    :type n: int
    :returns: la position finale (peut être négative).
    :rtype: int
    """
    position = 0
    i = 0
    while i < n:
        if randint(0, 1) == 0:
            position += 1
        else:
            position -= 1
        i += 1
    return position
```

```python
def test_marche():
    """Teste unitairement la fonction marche : invariants structurels 
    (une fonction aléatoire ne peut pas être testée sur une valeur exacte).

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    assert marche(0) == 0, "0 etape : la position doit rester 0"
    i = 0
    while i < 20:
        assert abs(marche(10)) <= 10, "la position ne peut pas depasser le nombre d'etapes"
        i += 1
    print("Test de la fonction marche : ok")

test_marche()
```

2. Exécuter cette fonction 10 000 fois avec `n = 100` et afficher la position moyenne obtenue (elle doit être proche de 0).

```python
#------------------------------#
#    CORRECTION  ! ! ! ! ! !   #
#------------------------------#
total = 0
nb_simulations = 10000
i = 0
while i < nb_simulations:
    total += marche(100)
    i += 1
print("Position moyenne :", total / nb_simulations)
```
