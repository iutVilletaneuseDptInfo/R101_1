# Modélisation des données (par listes)

Un livre est caractérisé par trois informations : son **ISBN**, son **titre** et son **auteur**. On le représente, dans cette première partie, par une liste à trois éléments :


````python
livre = ["1234567890", "Les Misérables", "Victor Hugo"]
````

- `livre[0]` est l'ISBN,
- `livre[1]` est le titre,
- `livre[2]` est l'auteur.

Le **catalogue** est alors une liste de livres, c'est-à-dire une liste de listes :


````python
catalogue = [
    ["1234567890", "Les Misérables", "Victor Hugo"],
    ["0987654321", "L'Étranger", "Albert Camus"],
]
````

Les **emprunts** en cours sont modélisés par une liste de chaînes de caractères `emprunts` telle que `emprunts[2*i]` contient le nom d'un lecteur et `emprunts[2*i+1]` contient l'ISBN du livre que ce lecteur a emprunté. Chaque paire consécutive d'éléments représente donc un emprunt en cours.

**Exemple** :


````python
emprunts = ["Alice", "1234567890", "Bob", "0987654321", "Bob", "1111111111"]
````
