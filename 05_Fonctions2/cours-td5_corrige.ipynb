## Fonctions avec une valeur de retour

Une fonction peut <em style="color:red">retourner</em> une valeur : elle la transmet alors au programme appelant, qui peut l'utiliser après l'appel de la fonction (dans un calcul, pour affecter une variable, *etc*.).

**Remarque :** Une fonction ne retourne qu'une seule valeur. Pour en retourner plusieurs, on utilise un tableau (*cf.* Chapitre 9) ou un tuple (non abordé dans ce cours).

### Définition

On retourne une valeur à l'aide de l'instruction `return`, suivie de la valeur à retourner.

```python
def nom_fonction(parametre1, parametre2, ..., parametrek) :
    # algorithme de la fonction
    
    return resultat_a_retourner
```

**Attention :**  L'instruction `return` termine l'exécution de la fonction. Elle peut être utilisée seule, sans valeur, simplement pour arrêter l'exécution (la fonction ne retourne dans ce cas aucune valeur).

### Appel

Après l'exécution d'un appel de fonction, cet appel est remplacé, dans le code du programme appelant, par la valeur retournée. Ainsi, pour récupérer la valeur retournée par `nom_fonction` et la stocker dans une variable `res`, on écrit :

```python
res = nom_fonction(valeur1, valeur2, ..., valeurk)
```

**Remarque :** la valeur retournée n'a pas besoin d'être affectée à une variable : elle peut aussi être affichée, utilisée dans un calcul, etc. Dans l'exemple suivant, la fonction `abs` est appelée deux fois, et les deux valeurs retournées sont additionnées puis affichées directement :

````python
print(abs(-2) + abs(-3))
````

### Exemple

````python
def perimetre_rectangle(longueur, largeur):
    """Retourne le périmètre d'un rectangle.

    :param longueur: la longueur du rectangle.
    :type longueur: float
    :param largeur: la largeur du rectangle.
    :type largeur: float
    :returns: le périmètre du rectangle.
    :rtype: float
    """
    perimetre = 2 * (longueur + largeur)
    return perimetre 
    #le résultat est transmis au programme appelant


pr = perimetre_rectangle(5.2, 3.4)
print("Le périmètre du rectangle est : ", pr)
````

Dans cet exemple, la fonction `perimetre_rectangle` 
- est appelée avec les valeurs 5.2 et 3.4 pour les paramètres `longueur` et `largeur`,
- calcule le périmètre et stocke sa valeur (17.2) dans la variable `perimetre`

Avec l'instruction `return`, elle retourne la valeur contenue dans `perimetre` (soit 17.2), puis les variables locales de la fonction et les paramètres sont détruites (soit `longueur`, `largeur` et `perimetre`).

Dans le programme appelant, la ligne 16 du code est évaluée comme : 
```python
pr = 17.2
```
car l'appel de la fonction a été remplacé par la valeur retournée.

**Remarque :** Exécuter le code avec `Python Tutor` permet de bien comprendre le processus.

**Remarque :** La valeur à retourner n'a pas besoin d'être stockée dans une variable pour être retournée. On aurait également pu définir la fonction `perimetre_rectangle` de cette manière :

````python
def perimetre_rectangle(longueur, largeur):
    """Retourne le périmètre d'un rectangle.

    :param longueur: la longueur du rectangle.
    :type longueur: float
    :param largeur: la largeur du rectangle.
    :type largeur: float
    :returns: le périmètre du rectangle.
    :rtype: float
    """
    return 2*(longueur + largeur)
````

### `None`

Il y a deux façons pour une fonction de se terminer sans transmettre de valeur utile au programme appelant : 
1. contenir l'instruction `return` seule (voir `somme_avec_return` ci-dessous), 

2. ou tout simplement ne pas contenir d'instruction `return` (voir `somme_sans_return`ci-dessous). 

Dans ces deux cas, la fonction ne renvoie pas « rien » : elle **renvoie une valeur bien précise**, `None`, qui représente en Python l'absence de valeur.

````python
def somme_avec_return(a, b):
    """Affiche la somme de deux nombres
    :param a: le premier nombre.
    :type a: int
    :param b: le second nombre.
    :type b: int
    :returns: None (affiche la somme, ne la retourne pas).
    :rtype: None
    """
    print(a + b)
    return

resultat = somme_avec_return(2,3) # l'exécution de la fonction affiche 5
print(resultat)  # None
````

**NB :** même en l'absence de l'instruction `return`, une fonction retourne par défaut la valeur `None`, comme le montre l'exemple suivant.

````python
def somme_sans_return(a, b):
    """Affiche la somme de deux nombres
    :param a: le premier nombre.
    :type a: int
    :param b: le second nombre.
    :type b: int
    :returns: None (affiche la somme, ne la retourne pas).
    :rtype: None
    """
    print(a + b)

resultat = somme_sans_return(2, 3)# l'exécution de la fonction affiche 5
print(resultat)  # None 
````

Pour qu'une fonction renvoie effectivement la valeur calculée, il ne faut donc pas oublier l'instruction `return`, comme dans la fonction `somme_calcul` ci-dessous :

````python
def somme_calcul(a,b):
    """Retourne la somme de deux nombres 

    :param a: le premier nombre.
    :type a: int
    :param b: le second nombre.
    :type b: int
    :returns: la somme de a et b.
    :rtype: int
    """
    s=a+b
    return s

resultat = somme_calcul(2,3) 
print(resultat) 
````

## Tests unitaires

Un programme contient généralement des bugs, et les développeurs passent beaucoup de temps à tester et déboguer leur code. Comme le nombre d'erreurs augmente avec la complexité du code, il est important de tester correctement chaque fonction. C'est ce que l'on appelle les *tests unitaires*.

Écrire des tests unitaires n'a rien d'évident : il faut envisager tous les cas d'usage de la fonction et vérifier qu'elle se comporte correctement dans chacun d'eux. Pour une fonction qui retourne une valeur, on l'appelle avec des entrées dont on connaît déjà le résultat attendu, et on vérifie que la valeur retournée correspond.

* Pour cela, on crée une fonction de test pour chaque fonction que l'on définit. Par convention, le nom de la fonction de test d'une fonction `nomFonction` est `test_nomFonction`. Elle ne prend aucun paramètre et ne retourne aucune valeur ;
* chaque test consiste à appeler la fonction avec des valeurs d'entrée pour lesquelles on connaît déjà le résultat attendu ;
* la comparaison entre la valeur attendue et la valeur retournée se fait avec l'instruction `assert` ;
* la fonction de test se termine par l'affichage de `Test de la fonction ... : ok` ;
* la fonction de test est définie et appelée dans une même cellule du notebook.

#### La fonction `assert`

- Syntaxe : `assert condition, message`
- Son comportement dépend de la condition :
  - **condition vraie** : rien ne se passe, le programme continue ;
  - **condition fausse** : le programme s'arrête et affiche `message`.

Voici un exemple de fonction de tests unitaires pour `perimetre_rectangle` :

````python
from math import *

# Fonction de tests unitaires de la fonction perimetre_rectangle
def test_perimetre_rectangle():
    assert perimetre_rectangle(5, 2) == 14
    assert perimetre_rectangle(0, 0) == 0, "erreur en 0,0"
    assert isclose(perimetre_rectangle(0.2, 0.1), 0.6), "erreur avec des floats"
    print("Test de la fonction perimetre_rectangle : ok")

test_perimetre_rectangle()
````

**Rappel :** Il faut absolument comparer les valeurs flottantes avec la fonction `math.isclose`. En effet, le test `perimetre_rectangle(0.2,0.1) == 0.6` aurait généré une erreur puisque `perimetre_rectangle(0.2,0.1)` est égal à 0.6000000000000001.

À chaque modification de la fonction `perimetre_rectangle`, on appelle `test_perimetre_rectangle` pour vérifier qu'elle satisfait toujours les tests unitaires.

**Important :** Tester des fonctions utilisant des saisies clavier ou des affichages écran est techniquement un peu complexe et ne sera pas vu dans ce cours. Pour tester ces fonctions, on écrira alors en commentaire la liste des saisies clavier à tester et/ou la liste des appels de fonction à tester en indiquant quels messages doivent être affichés à l'écran.

## Comment concevoir une fonction ?

### Démarche 

Lors de la définition d'une fonction, il est recommandé d'adopter la démarche suivante :

**Étape 1 :** *déterminer le rôle de la fonction.*

* Savoir précisément ce que fait la fonction et dans quel cadre elle sera utilisée.
* Lui donner un nom explicite.

**Étape 2 :** *déterminer les paramètres et la valeur de retour.*

* Écrire l'en-tête et le docstring.

**Étape 3 :** *écrire le corps de la fonction.*

**Étape 4 :** *écrire la fonction de tests unitaires.*

Les méthodes de développement TDD (*test driven development*) inversent l'ordre des étapes 3 et 4 : on écrit d'abord la fonction de tests unitaires, puis le corps de la fonction, de façon à ce que celui-ci soit guidé, dès le départ, par les tests qu'il doit satisfaire.

````python
def min_abs_intervalle(borne_min, borne_max):
    """Retourne l'entier de l'intervalle [borne_min, borne_max] 
    ayant la plus petite valeur absolue, s'il existe.

    :param borne_min: la borne inférieure de l'intervalle.
    :type borne_min: int
    :param borne_max: la borne supérieure de l'intervalle.
    :type borne_max: int
    :returns: l'entier de l'intervalle ayant la plus petite 
    valeur absolue, ou None si borne_min > borne_max.
    :rtype: int ou None
    """

    if borne_min > borne_max:
        val = None
    elif borne_min >= 0:
        val = borne_min
    elif borne_max > 0:
        val = 0
    else:
        val = -borne_max
    return val
````

**Exercice 5 : Série harmonique****

La série harmonique est la série définie pour tout $n >$ 0 par: 
$$H_n= 1 + \frac12 + \frac13 + \frac14 + \cdots + \frac1n$$

**Question 1**

Définir la fonction `harmonique` prenant en paramètre un entier n et retournant la valeur $H_n$ (la fonction retournera `-1` si $n$ est négatif ou nul).

````python
def harmonique(n):
    """Retourne la valeur Hn de la suite harmonique : 
    1 + 1/2 + 1/3 + ... + 1/n.

    :param n: le rang n de la suite harmonique.
    :type n: int
    :returns: la valeur de Hn si n > 0, -1 sinon.
    :rtype: float ou int
    """
    if n <= 0:
        return -1
     
    res = 0
    i=1
    while i <=n:
        res = res + 1/i
        i = i+1

    return res
````

**Question 2**

Définir une fonction de tests unitaire de la fonction `harmonique` sachant que : 
* $H_5 = 2.283333333333333$,
* $H_{20} = 3.597739657143682$.

**Attention :** La comparaison porte sur des flottants ici.

````python
#------------------#
#   CORRECTION !   #
#------------------#

from math import *

def test_harmonique():
    """Teste unitairement la fonction harmonique sur plusieurs valeurs de n, dont des cas invalides.

    :returns: None (échoue par assertion si un cas est incorrect).
    :rtype: None
    """
    assert isclose(harmonique(5), 2.283333333333333)
    assert isclose(harmonique(20), 3.597739657143682)
    assert harmonique(1)== 1
    assert harmonique(0)== -1
    assert harmonique(-10)== -1
    print("Test de la fonction harmonique : ok")

test_harmonique()
````

* Février contient 29 jours pour les années bissextiles et 28 jours sinon.
* Avril, juin, septembre et novembre contiennent 30 jours.

**Remarques :**
* La fonction retournera -1 si le mois n'est pas valide.
* On utilisera la fonction `estBissextile` pour déterminer si l'année passée en paramètre est bissextile.

````python
def nbjours(m, a):
    """Retourne le nombre de jours du mois m de l'année a.

    :param m: le numéro du mois (compris entre 1 et 12).
    :type m: int
    :param a: l'année.
    :type a: int
    :returns: le nombre de jours du mois m de l'année a, 
    ou -1 si m n'est pas un numéro de mois valide.
    :rtype: int
    """
    if m < 0 or m > 12:
        nbjours = -1
    elif m == 2:
        if estBissextile(a):
            nbjours = 29
        else:
            nbjours = 28
    elif m == 4 or m == 6 or m == 9 or m == 11:
        nbjours = 30
    else:
        nbjours = 31
    return nbjours
````

**Question 1**

Définir la fonction `jours1970` qui prend en paramètre une date et retourne le nombre de jours écoulés depuis le 1er janvier 1970. 

Si la date passée en paramètre n'est pas valide, la fonction retourne -1. 

Si la date est antérieure au 1er janvier 1970, la fonction retourne un nombre négatif, dont la valeur absolue est le nombre de jours séparant cette date du 01/01/1970.

````python
def jours1970(j, m, a):
    """Retourne le nombre de jours écoulés entre le 01/01/1970 
    et la date j/m/a.

    :param j: le jour.
    :type j: int
    :param m: le mois.
    :type m: int
    :param a: l'année.
    :type a: int
    :returns: le nombre de jours entre le 01/01/1970 et j/m/a 
    (négatif si la date est antérieure), ou -1 si j/m/a n'est pas
    une date valide.
    :rtype: int
    """
    if not estDateValide(j, m, a):
        return -1

    #Calcul du nombre de jours depuis le 1er janvier de l'année a
    nbj = 0
    i = 1
    while i < m:
        nbj += nbjours(i, a)
        i += 1
    nbj += j-1

    # Calcul du nombre de jours par année
    if a > 1970:
        i = 1971
        while i <= a:
            if estBissextile(i):
                nbj += 366
            else:
                nbj += 365
            i += 1
    elif a < 1970:
        i = a
        while i < 1970:
            if estBissextile(i):
                nbj -= 366
            else:
                nbj -= 365
            i += 1
    return nbj
````
