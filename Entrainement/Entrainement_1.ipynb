### Exercice : Affichage alternatif de `oui` ###
- Pour quelles valeurs de `x` s'affiche `oui` dans les
cas suivants ?

````python
if (x > 100):
    print("non")
else :
    print("oui")
````

<details>
<summary>&#128161; <b>Réponse</b></summary>

`oui` s'affiche pour toutes les valeurs `x <= 100` (le `else` couvre tout ce qui n'est pas `x > 100`).

</details>


````python
if (x > 100):
    print("non")
elif (x > 90):  #if a>90 and a <=100:
    print("oui")
````

<details>
<summary>&#128161; <b>Réponse</b></summary>

`oui` s'affiche pour `90 < x <= 100` (il faut échapper au premier `if`, donc `x <= 100`, et satisfaire le `elif`, donc `x > 90`).

</details>


````python
if (not(x > 0 and x < 100)):
    print("non")
elif (x > 70):
    print("oui")
````

<details>
<summary>&#128161; <b>Réponse</b></summary>

`oui` s'affiche pour `70 < x < 100` (il faut que `0 < x < 100` soit vrai pour échapper au `if`, et `x > 70` pour le `elif`).

</details>

- Que valent `a` et `b` à la fin du programme suivant ? 

````python
a = 1
b = 0
while a < 30 or b%5 == 0:
    a *= 2
    b += 7
print(a)
print(b)
````

<details>
<summary>&#128161; <b>Réponse</b></summary>

À la fin, `a` vaut `64` et `b` vaut `42` (la boucle continue tant que `a < 30` OU `b` est un multiple de 5 ; elle s'arrête dès que les deux conditions sont fausses en même temps).

</details>
