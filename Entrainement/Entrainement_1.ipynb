### Exercice : Affichage alternatif de `oui` ###
- Pour quelles valeurs de `x` s'affiche `oui` dans les
cas suivants ?

```python
if (x > 100):
    print("non")
else :
    print("oui")
```

**Réponse :**


```python
if (x > 100):
    print("non")
elif (x > 90): 
    print("oui")
```

**Réponse :**


```python
if (not(x > 0 and x < 100)):
    print("non")
elif (x > 70):
    print("oui")
```
