Dans un dictionnaire `d`, les valeurs sont toutes **différentes**. Écrire `inverse(d)` qui retourne le dictionnaire dont les clés sont les valeurs de `d` et dont les valeurs sont les clés correspondantes : `inverse({"a": 1, "b": 2})` vaut `{1: "a", 2: "b"}`.
