Pour chaque question, écrire la fonction demandée (avec sa docstring), puis l'appeler sur un exemple. Ces fonctions affichent un résultat mais ne retournent rien.

**a)** `affiche_ligne(n, symbole)` affiche `n` fois `symbole` sur une même ligne, puis passe à la ligne.

````python
def affiche_ligne(n, symbole):
    """Affiche un symbole répété n fois sur une ligne.

    :param n: nombre de répétitions
    :type n: int
    :param symbole: le symbole à afficher
    :type symbole: str
    """
    for i in range(n):
        print(symbole, end="")
    print()

affiche_ligne(5, "#")
````

**b)** `affiche_carres(n)` affiche, un par ligne, les carrés des entiers de 1 à `n`.

````python
def affiche_carres(n):
    """Affiche les carrés des entiers de 1 à n, un par ligne.

    :param n: dernier entier
    :type n: int
    """
    for i in range(1, n + 1):
        print(i * i)

affiche_carres(4)
````

**c)** `affiche_progression(debut, raison, nb)` affiche les `nb` premiers termes de la suite qui commence à `debut` et augmente de `raison` à chaque terme.

````python
def affiche_progression(debut, raison, nb):
    """Affiche les premiers termes d'une progression arithmétique.

    :param debut: premier terme
    :type debut: int
    :param raison: écart entre deux termes
    :type raison: int
    :param nb: nombre de termes
    :type nb: int
    """
    terme = debut
    for i in range(nb):
        print(terme, end=" ")
        terme = terme + raison
    print()

affiche_progression(3, 4, 6)   # 3 7 11 15 19 23
````

**d)** `affiche_vote(pour, contre)` affiche `adopté` si `pour > contre`, `rejeté` si `pour < contre` et `égalité` sinon.

````python
def affiche_vote(pour, contre):
    """Affiche le résultat d'un vote.

    :param pour: nombre de voix pour
    :type pour: int
    :param contre: nombre de voix contre
    :type contre: int
    """
    if pour > contre:
        print("adopté")
    elif pour < contre:
        print("rejeté")
    else:
        print("égalité")

affiche_vote(12, 9)
````

**e)** `affiche_somme(a, b)` affiche la somme de tous les entiers de `a` à `b` inclus (on suppose `a <= b`).

````python
def affiche_somme(a, b):
    """Affiche la somme des entiers de a à b inclus.

    :param a: premier entier
    :type a: int
    :param b: dernier entier
    :type b: int
    """
    s = 0
    for i in range(a, b + 1):
        s = s + i
    print(s)

affiche_somme(5, 10)   # 45
````

Écrire une fonction `affiche_ligne_barres(n)` qui affiche `n` barres `|` suivies, entre parenthèses, de la valeur de `n` (par exemple `|||| (4)`). En déduire une fonction `affiche_histogramme(a, b, c)` qui affiche trois lignes, une pour chacun de ses trois paramètres, en réutilisant la première fonction.

### Exercice 9 : Moyenne et médiane de trois nombres ★★ ###
Écrire deux fonctions : `moyenne3(a, b, c)` qui retourne la moyenne de trois nombres, et `mediane3(a, b, c)` qui retourne celui des trois qui est ni le plus petit ni le plus grand (la valeur « du milieu »). Les tests doivent passer quel que soit l'ordre des paramètres.

````python
def moyenne3(a, b, c):
    """Calcule la moyenne de trois nombres.

    :param a: premier nombre
    :type a: float
    :param b: deuxième nombre
    :type b: float
    :param c: troisième nombre
    :type c: float
    :returns: la moyenne de a, b et c
    :rtype: float
    """
    return (a + b + c) / 3

def mediane3(a, b, c):
    """Détermine la valeur médiane de trois nombres.

    :param a: premier nombre
    :type a: float
    :param b: deuxième nombre
    :type b: float
    :param c: troisième nombre
    :type c: float
    :returns: la valeur du milieu
    :rtype: float
    """
    if (a <= b and b <= c) or (c <= b and b <= a):
        return b
    elif (b <= a and a <= c) or (c <= a and a <= b):
        return a
    else:
        return c
````

````python
def test_moyenne3():
    assert moyenne3(3, 6, 9) == 6
    assert moyenne3(0, 0, 10) == 10 / 3
    print("Test de la fonction moyenne3 : ok")

test_moyenne3()

def test_mediane3():
    assert mediane3(1, 2, 3) == 2
    assert mediane3(3, 2, 1) == 2
    assert mediane3(2, 3, 1) == 2
    assert mediane3(5, 1, 3) == 3
    assert mediane3(4, 4, 1) == 4
    print("Test de la fonction mediane3 : ok")

test_mediane3()
````

````python
def est_triangle(a, b, c):
    """Teste si trois longueurs peuvent former un triangle.

    :param a: première longueur
    :type a: float
    :param b: deuxième longueur
    :type b: float
    :param c: troisième longueur
    :type c: float
    :returns: True si le triangle existe, False sinon
    :rtype: bool
    """
    return a < b + c and b < a + c and c < a + b
````

### Exercice 13 : Combien de chiffres ? ★★ ###
Écrire `nb_chiffres(n)` qui retourne le nombre de chiffres de l'écriture décimale de l'entier `n >= 1`, sans utiliser de chaîne de caractères.

````python
def nb_chiffres(n):
    """Compte les chiffres d'un entier strictement positif.

    :param n: un entier strictement positif
    :type n: int
    :returns: le nombre de chiffres de n
    :rtype: int
    """
    nb = 0
    while n > 0:
        n = n // 10
        nb = nb + 1
    return nb
````

````python
def test_nb_chiffres():
    assert nb_chiffres(7) == 1
    assert nb_chiffres(10) == 2
    assert nb_chiffres(20260) == 5
    assert nb_chiffres(999) == 3
    assert nb_chiffres(1000) == 4
    print("Test de la fonction nb_chiffres : ok")

test_nb_chiffres()
````

Indication : $\lfloor n / 10^k \rfloor \bmod 10$.

````python
def chiffre_de_rang(n, k):
    """Donne le chiffre de rang k d'un entier.

    :param n: un entier positif
    :type n: int
    :param k: rang du chiffre (0 pour les unités)
    :type k: int
    :returns: le chiffre de rang k (0 s'il n'existe pas)
    :rtype: int
    """
    return (n // 10 ** k) % 10
````

### Exercice 17 : Nombres amis ★★★ ###
Deux entiers `a` et `b` sont **amis** si la somme des diviseurs de `a` strictement inférieurs à `a` est égale à `b`, et réciproquement. Par exemple 220 et 284 sont amis.

Écrire `somme_diviseurs_stricts(n)` puis `sont_amis(a, b)` (qui doit utiliser la première fonction). On exclut le cas `a == b`.

````python
def somme_diviseurs_stricts(n):
    """Calcule la somme des diviseurs de n qui sont strictement inférieurs à n.

    :param n: un entier strictement positif
    :type n: int
    :returns: la somme des diviseurs stricts de n
    :rtype: int
    """
    s = 0
    for d in range(1, n):
        if n % d == 0:
            s = s + d
    return s

def sont_amis(a, b):
    """Teste si deux entiers sont amis.

    :param a: un entier strictement positif
    :type a: int
    :param b: un entier strictement positif
    :type b: int
    :returns: True si a et b sont amis, False sinon
    :rtype: bool
    """
    return a != b and somme_diviseurs_stricts(a) == b and somme_diviseurs_stricts(b) == a
````

````python
def test_somme_diviseurs_stricts():
    assert somme_diviseurs_stricts(1) == 0
    assert somme_diviseurs_stricts(12) == 16
    assert somme_diviseurs_stricts(220) == 284
    print("Test de la fonction somme_diviseurs_stricts : ok")

test_somme_diviseurs_stricts()

def test_sont_amis():
    assert sont_amis(220, 284)
    assert sont_amis(284, 220)
    assert not sont_amis(220, 285)
    assert not sont_amis(6, 6)
    print("Test de la fonction sont_amis : ok")

test_sont_amis()
````

### Exercice 18 : Premier multiple après n ★★ ###
Écrire `premier_multiple_apres(n, k)` qui retourne le plus petit multiple de `k` **strictement** supérieur à `n` (`k >= 1`, `n >= 0`), en utilisant une boucle `while`.

````python
def premier_multiple_apres(n, k):
    """Cherche le plus petit multiple de k strictement supérieur à n.

    :param n: un entier positif
    :type n: int
    :param k: un entier strictement positif
    :type k: int
    :returns: le plus petit multiple de k strictement supérieur à n
    :rtype: int
    """
    m = n + 1
    while m % k != 0:
        m = m + 1
    return m
````

````python
def test_premier_multiple_apres():
    assert premier_multiple_apres(10, 5) == 15
    assert premier_multiple_apres(11, 5) == 15
    assert premier_multiple_apres(0, 7) == 7
    assert premier_multiple_apres(20, 1) == 21
    print("Test de la fonction premier_multiple_apres : ok")

test_premier_multiple_apres()
````

````python
def annees_pour_doubler(taux):
    """Calcule le nombre d'années pour doubler un capital placé à intérêts composés.

    :param taux: taux annuel en pourcentage (strictement positif)
    :type taux: float
    :returns: le nombre d'années
    :rtype: int
    """
    capital = 1
    annees = 0
    while capital < 2:
        capital = capital * (1 + taux / 100)
        annees = annees + 1
    return annees
````

### Exercice 20 : Carré parfait ★★ ###
Écrire `est_carre_parfait(n)` qui retourne `True` si l'entier `n >= 0` est le carré d'un entier, sans utiliser de racine carrée.

````python
def est_carre_parfait(n):
    """Teste si un entier est un carré parfait.

    :param n: un entier positif
    :type n: int
    :returns: True si n est le carré d'un entier, False sinon
    :rtype: bool
    """
    i = 0
    while i * i < n:
        i = i + 1
    return i * i == n
````

````python
def test_est_carre_parfait():
    assert est_carre_parfait(0)
    assert est_carre_parfait(1)
    assert est_carre_parfait(16)
    assert not est_carre_parfait(15)
    assert not est_carre_parfait(17)
    assert est_carre_parfait(144)
    print("Test de la fonction est_carre_parfait : ok")

test_est_carre_parfait()
````

### Exercice 21 : Renverser un nombre, palindrome numérique ★★★ ###
Écrire `renverse(n)` qui retourne l'entier obtenu en lisant les chiffres de `n` à l'envers (`renverse(1230)` vaut `321`), puis `est_palindrome_nombre(n)` qui teste si `n` est égal à son renversé.

````python
def renverse(n):
    """Renverse l'ordre des chiffres d'un entier.

    :param n: un entier positif
    :type n: int
    :returns: l'entier écrit à l'envers
    :rtype: int
    """
    r = 0
    while n > 0:
        r = 10 * r + n % 10
        n = n // 10
    return r

def est_palindrome_nombre(n):
    """Teste si un entier se lit de la même façon dans les deux sens.

    :param n: un entier positif
    :type n: int
    :returns: True si n est un palindrome
    :rtype: bool
    """
    return n == renverse(n)
````

````python
def test_renverse():
    assert renverse(0) == 0
    assert renverse(7) == 7
    assert renverse(1234) == 4321
    assert renverse(1230) == 321
    print("Test de la fonction renverse : ok")

test_renverse()

def test_est_palindrome_nombre():
    assert est_palindrome_nombre(12321)
    assert est_palindrome_nombre(44)
    assert not est_palindrome_nombre(123)
    assert not est_palindrome_nombre(10)
    print("Test de la fonction est_palindrome_nombre : ok")

test_est_palindrome_nombre()
````

### Exercice 22 : Remplir une piscine ★★ ###
Écrire trois fonctions : `volume_pave(l, L, h)` (volume en m³ d'un pavé de dimensions `l`, `L`, `h` en mètres), `litres(volume)` (conversion de m³ en litres, 1 m³ = 1000 L), et `temps_remplissage(l, L, h, debit)` qui retourne le temps en heures pour remplir ce bassin avec un robinet de débit `debit` litres par heure. La dernière fonction doit réutiliser les deux autres.

````python
def volume_pave(l, L, h):
    """Calcule le volume d'un pavé droit.

    :param l: largeur en mètres
    :type l: float
    :param L: longueur en mètres
    :type L: float
    :param h: hauteur en mètres
    :type h: float
    :returns: le volume en m³
    :rtype: float
    """
    return l * L * h

def litres(volume):
    """Convertit un volume en m³ en litres.

    :param volume: volume en m³
    :type volume: float
    :returns: le volume en litres
    :rtype: float
    """
    return volume * 1000

def temps_remplissage(l, L, h, debit):
    """Calcule le temps de remplissage d'un bassin pavé droit.

    :param l: largeur en mètres
    :type l: float
    :param L: longueur en mètres
    :type L: float
    :param h: hauteur en mètres
    :type h: float
    :param debit: débit du robinet en litres par heure
    :type debit: float
    :returns: le temps de remplissage en heures
    :rtype: float
    """
    return litres(volume_pave(l, L, h)) / debit
````

````python
def test_volume_pave():
    assert volume_pave(2, 3, 4) == 24
    assert volume_pave(1, 1, 1) == 1
    print("Test de la fonction volume_pave : ok")

test_volume_pave()

def test_litres():
    assert litres(1) == 1000
    assert litres(0.5) == 500
    print("Test de la fonction litres : ok")

test_litres()

def test_temps_remplissage():
    assert temps_remplissage(2, 5, 1, 2500) == 4
    assert temps_remplissage(1, 1, 1, 500) == 2
    print("Test de la fonction temps_remplissage : ok")

test_temps_remplissage()
````

### Exercice 24 : Trouver le bug ★★ ###
La fonction `nb_multiples` ci-dessous doit compter les multiples de `k` compris entre `a` et `b` **inclus**. Exécuter les tests : ils échouent. Trouver l'erreur et écrire une version corrigée de la fonction.

````python
def nb_multiples(a, b, k):
    """Compte les multiples de k entre a et b inclus.

    :param a: borne inférieure
    :type a: int
    :param b: borne supérieure
    :type b: int
    :param k: un entier strictement positif
    :type k: int
    :returns: le nombre de multiples de k dans [a, b]
    :rtype: int
    """
    nb = 0
    for i in range(a, b):
        if i % k == 0:
            nb = nb + 1
    return nb
````

````python
def nb_multiples(a, b, k):
    """Compte les multiples de k entre a et b inclus.

    :param a: borne inférieure
    :type a: int
    :param b: borne supérieure
    :type b: int
    :param k: un entier strictement positif
    :type k: int
    :returns: le nombre de multiples de k dans [a, b]
    :rtype: int
    """
    nb = 0
    for i in range(a, b + 1):
        if i % k == 0:
            nb = nb + 1
    return nb
````

````python
def test_nb_multiples():
    assert nb_multiples(1, 10, 5) == 2
    assert nb_multiples(5, 5, 5) == 1
    assert nb_multiples(1, 4, 5) == 0
    assert nb_multiples(0, 20, 4) == 6
    print("Test de la fonction nb_multiples : ok")

test_nb_multiples()
````

L'erreur est classique : `range(a, b)` s'arrête à `b - 1`. Pour inclure `b`, il faut `range(a, b + 1)`. Les tests `nb_multiples(1, 10, 5)` et `nb_multiples(5, 5, 5)` détectent l'erreur car 10 et 5 sont des multiples situés sur la borne supérieure.
