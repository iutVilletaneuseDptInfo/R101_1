### Exercice 23 : Modifier alors qu'il ne le fallait pas ★★ ###
La fonction `avec_bonus` doit retourner un **nouveau** tableau où chaque note est augmentée de `bonus` (sans dépasser 20), et **laisser intact** le tableau des notes d'origine. Exécuter les tests : l'un d'eux échoue. Expliquer pourquoi, puis écrire une version corrigée.

````python
def avec_bonus(notes, bonus):
    """Ajoute un bonus à chaque note, avec un maximum de 20.

    :param notes: un tableau de notes
    :type notes: list
    :param bonus: le bonus à ajouter
    :type bonus: float
    :returns: le tableau des notes avec bonus
    :rtype: list
    """
    res = notes
    for i in range(len(res)):
        res[i] = res[i] + bonus
        if res[i] > 20:
            res[i] = 20
    return res
````

````python
#-------------------------#
#    Écrire le code ici   #
#-------------------------#


````

````python
def test_avec_bonus():
    notes = [10, 15, 19]
    assert avec_bonus(notes, 2) == [12, 17, 20]
    assert notes == [10, 15, 19]
    assert avec_bonus([], 3) == []
    print("Test de la fonction avec_bonus : ok")

test_avec_bonus()
````
