### Exercice 33 : Code secret ★★ ###
Le code secret d'une carte est `4217`. Écrire un programme qui demande le code (avec `input`, converti en entier) au plus 3 fois. Il affiche `Accès autorisé` dès que le code est correct, et `Carte bloquée` après trois échecs.

````python
secret = 4217
essais = 0
reussi = False
while essais < 3 and not reussi:
    code = int(input("Code : "))
    essais = essais + 1
    if code == secret:
        reussi = True
if reussi:
    print("Accès autorisé")
else:
    print("Carte bloquée")
````

La condition `essais < 3 and not reussi` arrête la boucle dans les deux cas : code correct ou trois essais épuisés.

Écrire un programme qui demande des nombres entiers à l'utilisateur jusqu'à ce qu'il saisisse `0` (qui n'est pas pris en compte). Le programme affiche ensuite la somme des nombres saisis et leur quantité.

````python
somme = 0
quantite = 0
n = int(input("Nombre (0 pour finir) : "))
while n != 0:
    somme = somme + n
    quantite = quantite + 1
    n = int(input("Nombre (0 pour finir) : "))
print(somme, quantite)
````

### Exercice 35 : Premier terme de Fibonacci dépassant 1000 ★★★ ###
Afficher le premier terme de la suite de Fibonacci qui dépasse strictement 1000, ainsi que son rang (le rang de $F_0$ est 0).

````python
a = 0
b = 1
rang = 0
while a <= 1000:
    a, b = b, a + b
    rang = rang + 1
print(a, rang)
````

### Exercice 36 : Chasse aux erreurs ★★ ###
Chaque programme contient une erreur. L'identifier et proposer une version corrigée.

**a)**

```python
# a) doit afficher 1 2 ... 10
i = 1
while i <= 10:
    print(i, end=" ")
```

**b)**

```python
# b) doit afficher 1 2 ... 10
for i in range(1, 10):
    print(i, end=" ")
```

**c)**

```python
# c) doit compter jusqu'à 5
n = 0
while n < 5:
    n + 1
print(n)
```

**d)**

```python
# d) doit afficher "cinq" si x vaut 5
if x = 5:
    print("cinq")
```

**Réponse :**

**a)** Boucle infinie : `i` n'est jamais modifié. Ajouter `i = i + 1` dans la boucle.

**b)** Erreur d'une unité : `range(1, 10)` s'arrête à 9. Il faut `range(1, 11)`.

**c)** `n + 1` calcule une valeur mais ne la range nulle part : boucle infinie. Il faut `n = n + 1`.

**d)** Pour tester l'égalité on écrit `==` ; `=` est l'affectation (erreur de syntaxe) : `if x == 5:`.

### Exercice 37 : Pairs et impairs ★ ###
Compter, avec deux variables, combien d'entiers pairs et combien d'entiers impairs il y a entre 1 et 57 (inclus). Afficher les deux résultats.

````python
pairs = 0
impairs = 0
i = 1
while i < 58:
    if i % 2 == 0:
        pairs = pairs + 1
    else:
        impairs = impairs + 1
    i = i + 1
print(pairs, impairs)
````

### Exercice 38 : Fizz Buzz ★★ ###
Pour chaque entier de 1 à 30, afficher `Fizz` s'il est multiple de 3, `Buzz` s'il est multiple de 5, `FizzBuzz` s'il est multiple de 3 et de 5, et l'entier lui-même sinon.

````python
i = 1
while i < 31:
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
    i = i + 1
````

L'ordre des tests est important : le cas « multiple de 3 **et** de 5 » doit être traité avant les deux autres.

### Exercice 39 : Diviseurs ★★ ###
La variable `n` vaut 36.

**a)** Afficher tous ses diviseurs.

````python
n = 36
d = 1
while d < n + 1:
    if n % d == 0:
        print(d, end=" ")
    d = d + 1
````

**b)** Afficher le nombre de ses diviseurs.

````python
n = 36
nb = 0
d = 1
while d < n + 1:
    if n % d == 0:
        nb = nb + 1
    d = d + 1
print(nb)
````

**c)** Afficher la somme de ses diviseurs strictement inférieurs à `n`.

````python
n = 36
s = 0
d = 1
while d < n:
    if n % d == 0:
        s = s + d
    d = d + 1
print(s)
````

### Exercice 40 : Nombre de chiffres ★★ ###
La variable `n` est un entier strictement positif (par exemple `n = 20260`). Afficher son nombre de chiffres en base 10, **sans** convertir `n` en chaîne.

Indication : combien de fois peut-on diviser `n` par 10 (division entière) avant d'arriver à 0 ?

````python
n = 20260
nb = 0
while n > 0:
    n = n // 10
    nb = nb + 1
print(nb)
````

### Exercice 41 : Renverser un nombre ★★★ ###
La variable `n` est un entier strictement positif (par exemple `n = 1234`). Afficher le nombre obtenu en écrivant ses chiffres à l'envers (`4321`), sans chaîne de caractères.

Indication : `n % 10` donne le dernier chiffre ; pour ajouter un chiffre `c` à droite d'un nombre `r`, on calcule `10 * r + c`.

````python
n = 1234
envers = 0
while n > 0:
    envers = 10 * envers + n % 10
    n = n // 10
print(envers)
````
